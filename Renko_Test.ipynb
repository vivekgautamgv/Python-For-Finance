```
import mplfinance as mpl
import pandas as pd
import yfinance as yf

## Data Gather
btc = yf.Ticker("BTC-USD").history(period = "1d", interval = "1m")

print(btc)



mpl.plot(
    btc,
    type="renko",
    renko_params=dict(brick_size=200),
    figscale=3.0 # Increase this number to enlarge the plot (default is 1.0)
)
```
